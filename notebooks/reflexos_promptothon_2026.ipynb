## 1. Exact work bypasses the LLM
A deterministic task should consume zero model calls and remain cacheable only after its acceptance checks pass.

```python
from reflexos import Engine, SimulatedProvider, Task
task = Task(task_id='exact-demo', objective='Count words exactly', operation='text_stats', input_data='AI routing should be inspectable', checks=({'kind':'json_equals','path':['words'],'value':5},))
result = Engine(SimulatedProvider()).run(task)
print('status:', result.status)
print('path:', result.path)
print('model_calls:', result.model_calls)
print('output:', result.output)
```

## 2. LUNA-FIRST escalation is executable policy
The simulator makes the path visible without credentials: difficulty 0 is solved by Luna, difficulty 1 requires Sol, and difficulty 2 requires Astra.

```python
def simulate(difficulty):
    task_id = f'demo-{difficulty}'
    t = Task(task_id=task_id, objective='Return the verified marker', checks=({'kind':'contains','value':f'PASS:{task_id}'},), metadata={'difficulty':difficulty,'expected':f'PASS:{task_id}'})
    return Engine(SimulatedProvider()).run(t)

for difficulty in (0,1,2):
    r = simulate(difficulty)
    print(difficulty, '=>', ' -> '.join(r.path), '| accepted:', r.status, '| calls:', r.model_calls)
```

These values isolate the orchestration policy. A real-provider evaluation must report provider/model versions, measured usage, latency and actual price separately.

## 4. Architecture principles

1. Deterministic tools first.
2. JEV semantic routing is optional and bounded.
3. Luna is the mandatory first generative tier.
4. Failed host-side acceptance checks are evidence for escalation.
5. Hard budgets and human checkpoints remain in code.
6. Provider failures never expose credentials in run telemetry.
